```python
from fastapi import FastAPI

app = FastAPI(title="BrandGuard API")

@app.get("/api/health")
def health_check():
    return {
        "status": "success",
        "message": "BrandGuard backend is working!"
    }
```

```python
import uvicorn
import threading

def run_server():
    uvicorn.run(app, host="127.0.0.1", port=8000)

server_thread = threading.Thread(target=run_server, daemon=True)
server_thread.start()
```
